# Ejercicio 1: Modelado de Interacciones Cotidianas mediante Autómatas Celulares ($S=\{A,R\}$)

## Contexto Teórico y Formalización
Desde la perspectiva de la sociología computacional y el modelado de sistemas complejos discretos, las dinámicas de interacción humana en entornos cotidianos pueden abstraerse formalmente mediante la teoría de **Autómatas Celulares (AC)**. Un sistema complejo discreto $S = \{A, R\}$ está constituido por un conjunto de agentes $A$ enlazados por un conjunto de relaciones o espacios de vecindad $R$.

Para cada uno de los tres escenarios analizados, representamos la dinámica micro-macro mediante la **tupla reducida $(w, U, f)$**, definida como:
* **$w$**: Alfabeto finito de estados discretos que puede asumir un individuo o celda $c_i$.
* **$U$**: Vecindad espacial o red social de radio $r$ que delimita el conjunto de celdas vecinas que influyen en el comportamiento local.
* **$f: w^{|U|} \rightarrow w$**: Regla local de transición que determina el estado de la celda en el instante $t+1$ con base en el estado de su vecindad en el instante $t$.

---

### Escenario 1: La Casa (Entorno Doméstico y Convivencia Familiar)

1. **Definición Formal de la Tupla Reducida $(w, U, f)$:**
   * **Alfabeto de Estados ($w$):**
     $$w = \{0: \text{Inactivo/Descanso (Habitación privada)}, 1: \text{Actividad Individual/Aislamiento}, 2: \text{Uso de Espacio Común (Cocina/Comedor/Sala)}\}$$
   * **Vecindad ($U$):**
     Vecindad basada en la topología de la vivienda con radio $r = 1$. Para la celda $c_i$, $U(i) = \{c_i\} \cup \{c_j \mid \text{distancia\_física}(c_i, c_j) \le 1\}$, que comprende las habitaciones o zonas contiguas.
   * **Reglas Locales de Transición ($f$):**   
   $c_i(t+1) = f(c_i(t), U(i)(t))$ =  \begin{cases}
     2 & \text{si } \sum_{j \in U(i)} \mathbb{I}(c_j(t) = 2) \ge 1 \text{ (Atracción social / Sincronización de comidas)}, \\
     0 & \text{si } t \in \text{Franja Nocturna} \lor \left(\forall j \in U(i)\setminus\{i\}, c_j(t) = 0\right) \text{ (Silencio/Descanso contagioso)}, \\
     1 & \text{en otro caso (Respeto al espacio privado o trabajo personal)}.
     \end{cases}

2. **Comportamiento Macro o Emergente:**
   * **Sincronización Espacio-Temporal y Ritmos Circadianos del Hogar:** Sin una autoridad centralizada que dicte horarios rígidos, las reglas micro de atenuación de ruido y atracción hacia zonas comunes producen la emergencia de un ritmo colectivo del hogar. Emergen patrones macro como horas pico autorreguladas de uso de zonas comunes (desayuno/cena) y períodos globales de quietud nocturna.

### Escenario 2: La Universidad / Aula de Clase

1. **Definición Formal de la Tupla Reducida $(w, U, f)$:**
   * **Alfabeto de Estados ($w$):**
     $$w = \{0: \text{Atención / Escucha Activa}, 1: \text{Distracción Local (Celular/Charla)}, 2: \text{Participación Académica}\}$$
   * **Vecindad ($U$):**
     Teselación bidimensional regular $2D$ correspondiente a la disposición de pupitres (Vecindad de Moore con radio $r = 1$). $U(i)$ contiene los 8 pupitres adyacentes a la celda $c_i$.
   * **Reglas Locales de Transición ($f$):**   
   $c_i(t+1) = \begin{cases}
     1 & \text{si } \sum_{j \in U(i)\setminus\{i\}} \mathbb{I}(c_j(t) = 1) \ge \theta_{\text{contagio}} \quad (\text{Efecto contagio de distracción/ruido}), \\
     0 & \text{si } \text{EstímuloDocente}(t) = \text{Alto} \land c_i(t) \neq 2 \quad (\text{Restablecimiento de atención}), \\
     2 & \text{si } \text{PreguntaAbierta}(t) \land \left(\sum_{j \in U(i)} \mathbb{I}(c_j(t) = 2) = 0\right) \quad (\text{Inhibición por participación cercana}).
     \end{cases}$

2. **Comportamiento Macro o Emergente:**
   * **Clusterización Social y Regímenes de Atención:** Se observan patrones emergentes de segregación espacial en el aula. Surgen "cúmulos de atención" (generalmente en las primeras filas) y "bolsones de distracción" (clusters en filas posteriores). El contagio norma-dependiente genera frentes de onda de murmullo o de atención concentrada que se propagan por el aula según la dinámica local del vecindario.

### Escenario 3: Medio de Transporte Público (Ej. Autobús o Transmilenio)

1. **Definición Formal de la Tupla Reducida $(w, U, f)$:**
   * **Alfabeto de Estados ($w$):**
     $$w = \{0: \text{Espacio Libre/Asiento Vacío}, 1: \text{Pasajero Estático/Aislamiento}, 2: \text{Pasajero en Movimiento/Flujo}\}$$
   * **Vecindad ($U$):**
     Red bidimensional acotada con restricciones de corredor/pasillo (Radio $r = 1$). Evalúa la densidad de contacto físico dentro del espacio vital inmediato (Burbuja Proxémica).
   * **Reglas Locales de Transición ($f$):**   
   $c_i(t+1)$ = \begin{cases}
     1 & \text{si } c_i(t) = 1 \land \text{Densidad}(U(i)) \le \text{CapacidadLímite}, \\
     2 & \text{si } \text{CercaDePuerta}(i) \land \text{EventoParada}(t) \quad (\text{Vector de salida}), \\
     0 & \text{si } c_i(t) = 2 \land \text{AvanceExitosoA}(i+1).
     \end{cases}
     *Regla Implícita de Proxémica:* Los individuos eligen celdas libres que maximicen la distancia euclidiana hacia las celdas ocupadas en estado $1$.

2. **Comportamiento Macro o Emergente:**
   * **Self-Organization y Auto-Segregación Proxémica:** A nivel macro emerge el llenado óptimo intercalado de asientos (ocupación de esquinas y asientos de ventana antes de pasillos), congelamiento de flujos por sobrepoblación (*jamming* o cuellos de botella en las puertas de salida durante las paradas) y fronteras de privacidad invisibles autorreguladas (evitación del contacto visual o uso de dispositivos personales como barrera social).
