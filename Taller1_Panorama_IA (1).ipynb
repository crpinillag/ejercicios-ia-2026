## Ejercicio 1.3: Análisis del Video (Vehículos de Braitenberg y Vida Artificial)

### 1. Resumen del Video y Concepto Central
El video expone los experimentos mentales descritos por Valentino Braitenberg en su obra *Vehicles: Experiments in Synthetic Psychology*. Muestra cómo agentes mecánicos simples con sensores acoplados directamente a motores (actuadores) generan comportamientos complejos que un observador externo calificaría de "inteligentes", "temerosos", "agresivos" o "exploradores", sin poseer un cerebro o modelo cognitivo explícito.

### 2. Conexión con el Capítulo de Lectura

* **Enfoque de Agentes y Sistemas Bio-inspirados:**
  El documento define a un agente inteligente como un ente que percibe el entorno a través de sensores y actúa mediante actuadores. Los Vehículos de Braitenberg demuestran el principio de la **Vida Artificial**: cómo la supervivencia y la adaptación surgen de interacciones mecánicas con el medio sin necesidad de lógica simbólica compleja.

* **Comportamiento Emergente en Sistemas Complejos:**
  En la Teoría General de Sistemas vista en el capítulo, un sistema se define como $S=\{A,R\}$. La dinámica del sistema produce una medida de comportamiento emergente $Q=f(q_1, q_2, \dots, q_n)$. En los vehículos de Braitenberg, la conducta "inteligente" $Q$ no está codificada en los componentes individuales (los sensores o ruedas $q_i$), sino que **emerge** del tipo de relación o cableado $R$ (conexiones cruzadas, inhibitorias o excitatorias) con el entorno.

* **De la Reacción Simple a la IA Física:**
  El video ejemplifica el salto conceptual hacia la IA Física expuesta por el profesor: máquinas que no solo procesan información digital, sino que incorporan sensorialidad y corporalidad para tomar decisiones y actuar dinámicamente en el mundo real en tiempo real.

## Ejercicio 1.4: Resumen Técnico sobre NVIDIA Cosmos

### 1. ¿Qué es NVIDIA Cosmos?
**NVIDIA Cosmos** es una plataforma de **Modelos Base del Mundo (*World Foundation Models - WFMs*)** diseñada aceleradamente para potenciar la IA Física, el desarrollo de robótica avanzada y la simulación de vehículos autónomos.

### 2. Características y Pilares Principales
* **Comprensión y Predicción del Mundo Físico:** A diferencia de los LLMs tradicionales orientados a texto o imágenes 2D, Cosmos está entrenado con millones de horas de datos de video físico y sensorial. Permite a las máquinas predecir leyes de la física, inercias, colisiones y dinámica de fluidos.
* **Generación de Entornos Sintéticos:** Capaz de generar simulaciones en video hiperrealistas para entrenar agentes autónomos en escenarios de alto riesgo o poco comunes (*edge cases*) sin poner en peligro equipos físicos.
* **Tokenización Multimodal Eficiente:** Utiliza tokenizadores de video y física que procesan estados espaciotemporales a alta velocidad, permitiendo razonamiento en tiempo real para robots y simulación en plataformas como NVIDIA Omniverse.

### 3. Relación con el Marco del Capítulo
NVIDIA Cosmos responde directamente al concepto de **IA Física** planteado en el texto. Al actuar en **sistemas complejos** (donde las variables del mundo real no se pueden modelar con ecuaciones lineales simples), Cosmos proporciona la base probabilística para que los agentes predigan las consecuencias de sus acciones antes de ejecutarlas mediante sus actuadores en el mundo real.

## Ejercicio 1.5: Planes del Gobierno de Colombia con respecto a la IA

### 1. Marco Estratégico Nacional (CONPES 3975 y Políticas Actuales)
El Gobierno Nacional de Colombia, a través del Ministerio de Tecnologías de la Información y las Comunicaciones (MINTIC) y el Departamento Nacional de Planeación (DNP), ha establecido hojas de ruta para la transformación digital y la adopción de IA:

* **Política Nacional para la Transformación Digital e Inteligencia Artificial (CONPES 3975):** Define las bases para promover el uso ético, legal y eficiente de la IA en el sector público y privado.
* **Estrategia Nacional de Datos y Hubs de IA:** Creación de ecosistemas regionales de innovación y laboratorios de IA (como las iniciativas en la Universidad Nacional y centros de supercómputo) para reducir la brecha tecnológica.

### 2. Ejes Fundamentales de Acción
1. **Infraestructura e IA Soberana:** Proyectos para dotar al país de capacidad de cómputo propia y centros de datos que mitiguen la dependencia tecnológica geopolítica planteada en el capítulo (como el acceso a hardware especializado).
2. **Ética y Gobernanza:** Hoja de ruta para el uso responsable de algoritmos en la toma de decisiones estatales, asegurando la transparencia y la protección de datos personales.
3. **Formación de Talento Humano:** Programas de democratización en Ciencia de Datos, Machine Learning y programación para preparar a la fuerza laboral ante la automatización de empleos.

### 3. Conexión con la Geopolítica Tecnológica
Como señala el documento de estudio, la IA ha dejado de ser solo una disciplina de ingeniería para convertirse en un elemento central de la geopolítica global. Los planes gubernamentales buscan evitar que el país quede rezagado como mero consumidor de APIs extranjeras, fomentando el desarrollo local de soluciones en IA para la agricultura, la salud y la gestión pública.
